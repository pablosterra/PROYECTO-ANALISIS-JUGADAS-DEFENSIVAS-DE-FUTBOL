```python
def dibujar_frame(frame):

    # Buscar el frame
    home = df_home[df_home['Frame'] == frame]
    away = df_away[df_away['Frame'] == frame]

    if home.empty or away.empty:
        print("El frame no existe.")
        return

    home = home.iloc[0]
    away = away.iloc[0]

    # Crear campo
    pitch = Pitch(
        pitch_type='custom',
        pitch_length=105,
        pitch_width=68,
        line_zorder=2
    )

    fig, ax = pitch.draw(figsize=(12, 8))


    # JUGADORES HOME

    for jugador in jugadores_home:

        x = home[jugador + '_X']
        y = home[jugador + '_Y']

        if pd.notna(x) and pd.notna(y):

            ax.scatter(
                x * 105,
                y * 68,
                s=120,
                marker='o',
                color='blue',
                label='HOME' if jugador == jugadores_home[0] else ""
            )


    # JUGADORES AWAY


    for jugador in jugadores_away:

        x = away[jugador + '_X']
        y = away[jugador + '_Y']

        if pd.notna(x) and pd.notna(y):

            ax.scatter(
                x * 105,
                y * 68,
                s=120,
                marker='^',
                color='red',
                label='AWAY' if jugador == jugadores_away[0] else ""
            )


    # BALÓN


    ball_x = home['Ball_X']
    ball_y = home['Ball_Y']

    if pd.notna(ball_x) and pd.notna(ball_y):

        ax.scatter(
            ball_x * 105,
            ball_y * 68,
            s=80,
            marker='*',
            color='orange',
            label='Balón'
        )

    # Título
    ax.set_title(
        f'Frame {frame} - Tiempo: {home["Time [s]"]:.2f} s'
    )

    ax.legend()

    plt.show()
```

función para encontrar el punto mas cercano a cada jugador

```python
def punto_mas_cercano(x_jugador, y_jugador, df_malla):

    distancias = np.sqrt(
        (df_malla['X'] - x_jugador)**2 +
        (df_malla['Y'] - y_jugador)**2
    )

    indice = distancias.idxmin()

    return df_malla.loc[indice, 'Punto']
```

```python
todos_los_jugadores = []

# Todos los frames del partido
frame_inicial = df_home['Frame'].min()
frame_final = df_home['Frame'].max()

for frame in range(frame_inicial, frame_final + 1):

    # Buscar el frame
    home = df_home[df_home['Frame'] == frame]
    away = df_away[df_away['Frame'] == frame]

    if home.empty or away.empty:
        continue

    home = home.iloc[0]
    away = away.iloc[0]

    jugadores_frame = []

    # JUGADORES HOME

    for jugador in jugadores_home:

        x = home[jugador + '_X']
        y = home[jugador + '_Y']

        if pd.notna(x) and pd.notna(y):

            jugadores_frame.append({
                'Frame': frame,
                'Jugador': jugador,
                'Equipo': 'HOME',
                'X': x,
                'Y': y
            })

    # JUGADORES AWAY

    for jugador in jugadores_away:

        x = away[jugador + '_X']
        y = away[jugador + '_Y']

        if pd.notna(x) and pd.notna(y):

            jugadores_frame.append({
                'Frame': frame,
                'Jugador': jugador,
                'Equipo': 'AWAY',
                'X': x,
                'Y': y
            })

    todos_los_jugadores.extend(jugadores_frame)

# Crear DataFrame final
df_jugadores_partido = pd.DataFrame(todos_los_jugadores)

df_jugadores_partido
```

```python
posiciones_eventos = []

for indice, evento in df_goal_kicks.iterrows():

    frame_inicio = evento['Start Frame Analisis']
    frame_fin = evento['End Frame']

    # Buscar todos los jugadores entre el inicio y final del evento
    posiciones = df_jugadores_partido[
        df_jugadores_partido['Frame'].between(
            frame_inicio,
            frame_fin
        )
    ].copy()

    posiciones['Evento'] = indice
    posiciones['Subtype'] = evento['Subtype']
    posiciones['Equipo_evento'] = evento['Team']

    posiciones_eventos.append(posiciones)



df_jugadores_eventos = pd.concat(
    posiciones_eventos,
    ignore_index=True
)

df_jugadores_eventos
```

Creo una tabla de caracteristicas

```python
caracteristicas = []

for (evento, frame, equipo), grupo in df_jugadores_eventos.groupby(
    ['Evento', 'Frame', 'Equipo']):

    # Centro del equipo
    centro_x = grupo['X'].mean()
    centro_y = grupo['Y'].mean()

    # Longitud del equipo
    longitud = grupo['X'].max() - grupo['X'].min()

    # Anchura del equipo
    anchura = grupo['Y'].max() - grupo['Y'].min()

    # Distancia de cada jugador al centro del equipo
    distancias_centro = np.sqrt(
        (grupo['X'] - centro_x)**2 +
        (grupo['Y'] - centro_y)**2)

    # Dispersión media
    dispersion = distancias_centro.mean()

    caracteristicas.append({
        'Evento': evento,
        'Frame': frame,
        'Equipo': equipo,
        'Centro_X': centro_x,
        'Centro_Y': centro_y,
        'Anchura': anchura,
        'Longitud': longitud,
        'Dispersion': dispersion})


df_caracteristicas = pd.DataFrame(caracteristicas)

df_caracteristicas
```

```python
def dibujar_frame_con_malla(frame):

    home = df_home[df_home['Frame'] == frame]
    away = df_away[df_away['Frame'] == frame]

    if home.empty or away.empty:
        print("El frame no existe.")
        return

    home = home.iloc[0]
    away = away.iloc[0]

    pitch = Pitch(
        pitch_type='custom',
        pitch_length=105,
        pitch_width=68,
        line_zorder=2
    )

    fig, ax = pitch.draw(figsize=(12, 8))

    # MALLA

    df_malla['X']
    df_malla['Y']

    ax.scatter(
        X_malla,
        Y_malla,
        s=5,
        marker='.',
        alpha=0.5,
        label='Malla'
    )


    # JUGADORES HOME


    for jugador in jugadores_home:

        x = home[jugador + '_X']
        y = home[jugador + '_Y']

        if pd.notna(x) and pd.notna(y):

            ax.scatter(
                x * 105,
                y * 68,
                s=120,
                marker='o',
                color='blue',
                label='HOME' if jugador == jugadores_home[0] else ""
            )

    # JUGADORES AWAY


    for jugador in jugadores_away:

        x = away[jugador + '_X']
        y = away[jugador + '_Y']

        if pd.notna(x) and pd.notna(y):

            ax.scatter(
                x * 105,
                y * 68,
                s=120,
                marker='^',
                color='red',
                label='AWAY' if jugador == jugadores_away[0] else ""
            )


    # BALÓN

    ball_x = home['Ball_X']
    ball_y = home['Ball_Y']

    if pd.notna(ball_x) and pd.notna(ball_y):

        ax.scatter(
            ball_x * 105,
            ball_y * 68,
            s=100,
            marker='*',
            color='orange',
            label='Balón'
        )

    # TÍTULO

    ax.set_title(
        f'Frame {frame} - Tiempo: {home["Time [s]"]:.2f} s'
    )

    ax.legend()

    plt.show()



```

```python
eventos_home_p1 = df_tiros[
    (df_tiros['Period'] == 1) &
    (df_tiros['Team'] == 'Away')
].index

print("Tiros de Away en la primera parte:")
print(len(eventos_home_p1))

eventos_home_p1
```

```python
df_home_defensa_p1 = df_jugadores_tiros[
    (df_jugadores_tiros['Evento'].isin(eventos_home_p1)) &
    (df_jugadores_tiros['Equipo'] == 'HOME')
].copy()

print(
    "Jugadas analizadas:",
    df_home_defensa_p1['Evento'].nunique()
)

print(
    "Frames analizados:",
    df_home_defensa_p1['Frame'].nunique()
)
```

```python
caracteristicas_home_p1 = []

for (evento, frame), grupo in df_home_defensa_p1.groupby(
    ['Evento', 'Frame']
):

    # Centro del bloque defensivo
    centro_x = grupo['X'].mean()
    centro_y = grupo['Y'].mean()

    # Anchura del bloque
    anchura = grupo['Y'].max() - grupo['Y'].min()

    # Longitud del bloque
    longitud = grupo['X'].max() - grupo['X'].min()

    # Distancia de cada jugador al centro del bloque
    distancias_centro = np.sqrt(
        (grupo['X'] - centro_x)**2 +
        (grupo['Y'] - centro_y)**2
    )

    # Dispersión media
    dispersion = distancias_centro.mean()

    caracteristicas_home_p1.append({
        'Evento': evento,
        'Frame': frame,
        'Centro_X': centro_x,
        'Centro_Y': centro_y,
        'Anchura': anchura,
        'Longitud': longitud,
        'Dispersion': dispersion
    })


df_caracteristicas_home_p1 = pd.DataFrame(
    caracteristicas_home_p1
)

df_caracteristicas_home_p1
```

```python
from sklearn.preprocessing import StandardScaler

variables = [
    'Centro_X',
    'Centro_Y',
    'Anchura',
    'Longitud',
    'Dispersion'
]

scaler_home_p1 = StandardScaler()

X_home_p1 = scaler_home_p1.fit_transform(
    df_caracteristicas_home_p1[variables]
)
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inercias = []

for k in range(1, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_home_p1)
    inercias.append(kmeans.inertia_)


plt.figure(figsize=(8, 5))

plt.plot(
    range(1, 11),
    inercias,
    marker='o'
)

plt.xlabel('Número de clusters (k)')
plt.ylabel('Inercia')
plt.title('Método del codo - HOME defensa - Primera parte')
plt.xticks(range(1, 11))

plt.show()
```

```python
kmeans_home_p1 = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

df_caracteristicas_home_p1['Cluster'] = (
    kmeans_home_p1.fit_predict(X_home_p1)
)
```

```python
df_caracteristicas_home_p1['Cluster'].value_counts().sort_index()
```

```python
resumen_home_p1 = (
    df_caracteristicas_home_p1
    .groupby('Cluster')[variables]
    .mean()
    .round(3)
)

resumen_home_p1
```
